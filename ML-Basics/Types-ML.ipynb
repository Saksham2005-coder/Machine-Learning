# Types of Machine Learning

Machine Learning can be classified into **four major types** based on how the model learns from data.

1. Supervised Learning
2. Semi-Supervised Learning
3. Unsupervised Learning
4. Reinforcement Learning

---

# Overview

| Type | Uses Labeled Data? | Goal | Example |
|------|---------------------|------|---------|
| Supervised Learning | ✅ Yes | Predict output | Spam Detection |
| Semi-Supervised Learning | ⚠️ Partially | Learn from few labels + many unlabeled samples | Medical Image Classification |
| Unsupervised Learning | ❌ No | Find hidden patterns | Customer Segmentation |
| Reinforcement Learning | ❌ No Labels | Learn by rewards and penalties | Self-Driving Cars |

---

# 1. Supervised Learning

## Definition

**Supervised Learning** is a type of Machine Learning where the model is trained using **labeled data**.

In labeled data, each input has a corresponding **correct output (label)**.

The model learns the relationship between the input and the output and then predicts the output for new, unseen data.

> **Simple Definition:**  
> Supervised Learning is learning with the help of a teacher, where the correct answers are already known.

---

## Real-Life Analogy

Imagine a teacher teaching students mathematics.

```
Question → Correct Answer
```

The student learns from these examples and can solve similar questions in the future.

Similarly, a Machine Learning model learns from labeled examples.

---

## Working Process

```
Labeled Data
(Input + Output)
        │
        ▼
Train Machine Learning Model
        │
        ▼
Model Learns Relationship
        │
        ▼
Predict Output for New Data
```

---

## Example

### House Price Prediction

Training Data

| Area (sqft) | Price (₹ Lakh) |
|--------------|----------------|
| 1000 | 40 |
| 1500 | 60 |
| 2000 | 80 |

The model learns:

> Larger house → Higher price

Now if we give:

```
Area = 1800 sqft
```

The model predicts:

```
Price ≈ ₹72 Lakh
```

---

## Types of Supervised Learning

### 1. Regression

Predicts continuous numerical values.

Examples

- House Price Prediction
- Temperature Prediction
- Sales Forecasting

Output Example

```
₹52.5 Lakh
```

---

### 2. Classification

Predicts categories or classes.

Examples

- Spam / Not Spam
- Cancer / No Cancer
- Cat / Dog

Output Example

```
Spam
```

---

## Popular Algorithms

- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)
- Naive Bayes

---

## Applications

- Disease Prediction
- Fraud Detection
- Email Spam Detection
- House Price Prediction
- Face Recognition
- Sentiment Analysis

---

## Advantages

- High accuracy with quality labeled data
- Easy to evaluate
- Well-suited for prediction problems

---

## Disadvantages

- Requires labeled data
- Labeling data can be expensive and time-consuming
- Performance depends on data quality

---

# 2. Semi-Supervised Learning

## Definition

**Semi-Supervised Learning** combines **Supervised Learning** and **Unsupervised Learning**.

The model is trained using:

- A **small amount of labeled data**
- A **large amount of unlabeled data**

This is useful when labeling data is difficult, costly, or time-consuming.

> **Simple Definition:**  
> Semi-Supervised Learning uses a few labeled examples to help learn from many unlabeled examples.

---

## Real-Life Analogy

Imagine a teacher labels only **10 questions** in a textbook.

The student uses those solved examples to understand and solve the remaining **990 questions**.

---

## Working Process

```
Small Labeled Data
        │
        ▼
Large Unlabeled Data
        │
        ▼
Model Learns from Both
        │
        ▼
Improved Prediction
```

---

## Example

Suppose a hospital has:

- 1,000 labeled X-ray images
- 100,000 unlabeled X-ray images

Instead of labeling all 100,000 images, the model learns from the small labeled set and improves using the large unlabeled set.

---

## Applications

- Medical Image Classification
- Speech Recognition
- Face Recognition
- Web Page Classification
- Document Classification

---

## Advantages

- Requires fewer labeled samples
- Lower labeling cost
- Better performance than using only a small labeled dataset

---

## Disadvantages

- More complex training process
- Quality of unlabeled data affects performance
- Sensitive to incorrect labels

---

# 3. Unsupervised Learning

## Definition

**Unsupervised Learning** is a type of Machine Learning where the model is trained using **unlabeled data**.

The model does **not know the correct answers**.

Instead, it discovers hidden patterns, similarities, or structures in the data.

> **Simple Definition:**  
> Unsupervised Learning finds hidden patterns in data without being told the correct answers.

---

## Real-Life Analogy

Imagine entering a room full of people you have never met.

Without anyone telling you, you naturally group people based on similarities such as:

- Age
- Profession
- Language
- Clothing

This is similar to Unsupervised Learning.

---

## Working Process

```
Unlabeled Data
       │
       ▼
Machine Learning Algorithm
       │
       ▼
Find Hidden Patterns
       │
       ▼
Groups / Clusters / Relationships
```

---

## Example

Customer Segmentation

A shopping website groups customers based on:

- Purchase history
- Age
- Spending habits

Without knowing customer categories beforehand, the model automatically forms groups.

---

## Main Tasks

### 1. Clustering

Groups similar data points together.

Examples

- Customer Segmentation
- Friend Suggestions
- Market Analysis

Algorithms

- K-Means
- DBSCAN
- Hierarchical Clustering

---

### 2. Association

Finds relationships between items.

Example

People who buy:

```
Bread
Milk
```

often also buy

```
Butter
```

Used in supermarket recommendation systems.

Algorithm

- Apriori

---

### 3. Dimensionality Reduction

Reduces the number of features while keeping important information.

Examples

- PCA (Principal Component Analysis)
- t-SNE
- UMAP

---

## Applications

- Customer Segmentation
- Recommendation Systems
- Market Basket Analysis
- Anomaly Detection
- Data Compression

---

## Advantages

- No labeled data required
- Finds hidden structures
- Useful for exploratory data analysis

---

## Disadvantages

- Difficult to evaluate
- Results may not always be meaningful
- Interpretation can be challenging

---

# 4. Reinforcement Learning

## Definition

**Reinforcement Learning (RL)** is a type of Machine Learning where an **agent learns by interacting with an environment**.

The agent performs actions and receives:

- **Reward** for correct actions
- **Penalty** for incorrect actions

The goal is to maximize the total reward over time.

> **Simple Definition:**  
> Reinforcement Learning teaches an agent through trial and error using rewards and penalties.

---

## Real-Life Analogy

Teaching a dog tricks.

- Correct action → Give a treat (reward)
- Wrong action → No treat (penalty)

Over time, the dog learns the desired behavior.

Similarly, an RL agent learns the best actions through experience.

---

## Components of Reinforcement Learning

```
Agent
   │
Chooses Action
   │
Environment
   │
Returns Reward
   │
Agent Learns
```

---

## Working Process

```
Current State
      │
      ▼
Agent Takes Action
      │
      ▼
Environment Responds
      │
      ▼
Reward / Penalty
      │
      ▼
Agent Updates Strategy
      │
      ▼
Repeat
```

---

## Example

Self-Driving Car

State

```
Current road conditions
```

Action

```
Turn Left
Brake
Accelerate
```

Reward

```
Safe driving → +10

Collision → -100
```

The car gradually learns the safest driving strategy.

---

## Applications

- Self-Driving Cars
- Robotics
- Game Playing (Chess, Go, Atari)
- Drone Navigation
- Recommendation Systems
- Resource Management

---

## Popular Algorithms

- Q-Learning
- SARSA
- Deep Q Network (DQN)
- Policy Gradient
- PPO (Proximal Policy Optimization)

---

## Advantages

- Learns continuously through interaction
- Suitable for dynamic environments
- Can achieve optimal strategies over time

---

## Disadvantages

- Requires many training iterations
- Computationally expensive
- Training can be slow
- Reward design is challenging

---

# Comparison of All Four Types

| Feature | Supervised | Semi-Supervised | Unsupervised | Reinforcement |
|---------|------------|-----------------|--------------|---------------|
| Labeled Data | ✅ Yes | ⚠️ Partial | ❌ No | ❌ No |
| Unlabeled Data | ❌ No | ✅ Yes | ✅ Yes | Environment States |
| Teacher Available | Yes | Partially | No | Reward Signal |
| Goal | Predict output | Improve learning with few labels | Discover hidden patterns | Learn best actions |
| Output | Labels or values | Labels or values | Clusters, associations | Optimal policy |
| Example | Spam Detection | Medical Imaging | Customer Segmentation | Self-Driving Cars |

### Supervised Learning

```
Teacher gives questions and answers.
```

✅ Learn from labeled data.

---

### Semi-Supervised Learning

```
Teacher solves only a few questions.
```

✅ Learn from both labeled and unlabeled data.

---

### Unsupervised Learning

```
No teacher.
```

✅ Find hidden patterns on your own.

---

### Reinforcement Learning

```
Learn by trial and error.
```

✅ Rewards for good actions and penalties for bad actions.

# Interview / Exam Points

- **Supervised Learning** uses labeled data and is mainly used for **classification** and **regression** problems.
- **Semi-Supervised Learning** combines a small amount of labeled data with a large amount of unlabeled data.
- **Unsupervised Learning** works with unlabeled data to discover hidden patterns, clusters, and relationships.
- **Reinforcement Learning** trains an agent through interaction with an environment using rewards and penalties.
- Common supervised algorithms include **Linear Regression, Decision Tree, Random Forest, SVM, and KNN**.
- Common unsupervised algorithms include **K-Means, DBSCAN, Hierarchical Clustering, and PCA**.
- Common reinforcement learning algorithms include **Q-Learning, DQN, PPO, and SARSA**.

---

# Summary

Machine Learning is broadly divided into **Supervised Learning, Semi-Supervised Learning, Unsupervised Learning, and Reinforcement Learning**. Supervised Learning uses labeled data to predict outputs, Semi-Supervised Learning combines a small labeled dataset with a large unlabeled dataset, Unsupervised Learning discovers hidden patterns without labels, and Reinforcement Learning enables an agent to learn through rewards and penalties while interacting with an environment. Each type is suited for different kinds of real-world problems, from spam detection and customer segmentation to robotics and autonomous driving.
