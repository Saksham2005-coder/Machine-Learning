# Framing the Problem in Machine Learning

## Introduction

**Problem Framing** is the **first and most important step** in the Machine Learning Development Life Cycle (MLDLC).

Before collecting data or choosing an algorithm, we must clearly understand **what problem we are trying to solve**.

A poorly framed problem often leads to:

- Wrong data collection
- Wrong Machine Learning algorithm
- Poor predictions
- Wasted time and resources

> **Simple Definition:**  
> Problem Framing is the process of converting a real-world business problem into a Machine Learning problem.

---

# Why is Problem Framing Important?

Machine Learning is **not about using fancy algorithms**.

It is about solving a **real business problem**.

If the problem is not defined correctly, even the most advanced ML model will fail.

---

## Example

Business says:

```
"Our customers are leaving."
```

This is **not** an ML problem.

We need to convert it into something measurable.

Machine Learning Problem

```
Predict whether a customer will leave within the next 30 days.
```

Now the problem becomes clear.

---

# Steps in Problem Framing

```
Business Problem
        │
        ▼
Understand Business Goal
        │
        ▼
Convert into ML Problem
        │
        ▼
Choose Problem Type
        │
        ▼
Collect Data
        │
        ▼
Select Evaluation Metric
        │
        ▼
Choose Learning Method
        │
        ▼
Check Assumptions
```

# Case Study: Netflix Customer Churn Prediction

## Business Problem

Netflix earns money through monthly subscriptions.

Some users stop using Netflix after a few months.

This is called **Customer Churn**.

---

## What is Churn?

**Churn** means a customer **stops using a company's service**.

Examples

- Cancels subscription
- Stops purchasing
- Uninstalls the app

---

## Business Goal

Netflix wants to reduce customer churn because:

- Acquiring new customers is expensive.
- Retaining existing customers is cheaper.
- Lower churn increases profit.

---

## Current Situation

Suppose

```
Monthly Subscribers

10 Million
```

Every month

```
5%

customers cancel.
```

Monthly Churn

```
500,000 users
```

This causes huge revenue loss.

---

## Business Question

```
Can we identify customers who are likely to leave?
```

---

## Machine Learning Solution

Train a model to predict:

```
Will this customer leave next month?

Yes

or

No
```

---

# Online Learning vs Batch Learning

Another important decision is deciding **how the model should learn**.

---

## Batch Learning

Train the model once using all available data.

```
Train

↓

Deploy

↓

Retrain after several weeks/months
```

Suitable when:

- Data changes slowly.
- Large retraining intervals are acceptable.

---

## Online Learning

Continuously updates the model whenever new data arrives.

```
New Data

↓

Update Model

↓

Repeat
```

Suitable when:

- Data changes rapidly.
- Real-time adaptation is required.

---

## Netflix Example

Suppose user preferences change every day.

An Online Learning system can continuously update recommendations.

However, churn prediction is often retrained periodically using Batch Learning because churn labels are not available instantly.

---

## Comparison

| Batch Learning | Online Learning |
|----------------|-----------------|
| Learns once | Learns continuously |
| Needs retraining | Updates automatically |
| Stable | Adapts quickly |
| High memory | Low memory |
| Best for static data | Best for streaming data |

# Check Assumptions

Before starting the project, verify important assumptions.

---

## Questions to Ask

### Is enough data available?

Example

```
Need at least several thousand customer records.
```

---

### Is the data representative?

Does it include:

- New users?
- Old users?
- Different countries?
- Different subscription plans?

---

### Are labels available?

For churn prediction,

Need:

```
Customer Left?

Yes

No
```

---

### Can the problem actually be solved using ML?

Some problems are better solved using:

- Simple rules
- SQL queries
- Mathematical formulas

instead of Machine Learning.

---

### Is the expected accuracy realistic?

No model is 100% accurate.

Set realistic expectations.

Example

```
Target Accuracy

90%
```

# Interview / Exam Points

- **Problem Framing** is the first step in the Machine Learning Development Life Cycle (MLDLC).
- It converts a **business problem** into a **Machine Learning problem**.
- Before building a model, identify:
  - Business objective
  - ML problem type
  - Current solution
  - Data sources
  - Evaluation metrics
  - Learning approach (Online vs Batch)
  - Project assumptions
- In the Netflix case study, **customer churn prediction** is formulated as a **Binary Classification** problem.
- Choosing appropriate metrics (e.g., Precision, Recall, F1-Score, ROC-AUC for classification) is essential for evaluating model performance.

---

# Summary

Problem Framing is the foundation of every successful Machine Learning project. It involves understanding the business objective, translating it into an appropriate ML problem, selecting the correct problem type, analyzing the existing solution, gathering suitable data, defining evaluation metrics, deciding between Batch and Online Learning, and validating key assumptions. Proper problem framing ensures that the resulting Machine Learning model addresses real business needs and delivers meaningful, measurable value.
