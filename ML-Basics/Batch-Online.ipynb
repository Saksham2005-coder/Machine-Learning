# Batch Learning, Online Learning, and Out-of-Core Learning

Machine Learning models can also be classified based on **how they learn from data over time**.

The three major learning approaches are:

1. Batch Learning (Offline Learning)
2. Online Learning
3. Out-of-Core Learning

These approaches differ in **when and how the model is trained**, **how it handles new data**, and **the amount of memory required**.

---

# Overview

| Learning Type | Training Data | Model Updates | Memory Requirement | Best For |
|---------------|---------------|---------------|--------------------|----------|
| Batch Learning | Entire dataset at once | No (retrain required) | High | Small to medium datasets |
| Online Learning | One sample or small batches | Continuous | Low | Streaming or continuously changing data |
| Out-of-Core Learning | Small chunks of very large datasets | Batch-by-batch | Low | Datasets too large to fit into RAM |

---

# 1. Batch Learning (Offline Learning)

## Definition

**Batch Learning** is a learning approach in which the model is trained using the **entire training dataset at one time**.

Once training is complete, the model is deployed and **does not learn from new incoming data automatically**.

If new data becomes available, the model must be **trained again from scratch** (or retrained using the updated dataset).

> **Simple Definition:**  
> Batch Learning means learning everything at once from the complete dataset.

---

## Real-Life Analogy

Imagine studying for an exam.

You study **all chapters completely**, take the exam, and stop studying.

If a new chapter is added later, you must study everything again before the next exam.

This is exactly how Batch Learning works.

---

## Working Process

```
Complete Dataset
       │
       ▼
Train Model Once
       │
       ▼
Deploy Model
       │
       ▼
Prediction

(New Data Arrives)
       │
       ▼
Retrain Entire Model
```

---

## Example

Suppose we have:

```
100,000 customer records
```

The model trains using all 100,000 records.

One month later:

```
10,000 new records arrive.
```

The model **cannot learn automatically**.

Instead, we combine:

```
110,000 records
```

and train the model again.

---

## Characteristics

- Uses the complete dataset.
- Training usually takes longer.
- High memory requirement.
- No continuous learning.
- Requires retraining when new data arrives.

---

## Advantages

- Usually achieves high accuracy.
- Stable training process.
- Easier to optimize and debug.
- Suitable when data changes infrequently.

---

## Disadvantages

- Cannot adapt automatically to new data.
- Retraining is expensive.
- High RAM requirement.
- Not suitable for real-time applications.

---

## Applications

- House Price Prediction
- Student Result Prediction
- Disease Prediction
- Image Classification
- Offline Recommendation Systems

---

# 2. Online Learning

## Definition

**Online Learning** is a learning approach where the model learns **continuously** as new data arrives.

Instead of training on the entire dataset, the model updates itself using:

- One data sample at a time, or
- Small batches of data (mini-batches)

The model continuously improves without retraining from scratch.

> **Simple Definition:**  
> Online Learning means learning continuously from new incoming data.

---

## Real-Life Analogy

Imagine learning a new language.

Every day you learn a few new words.

You do not restart from the beginning each time.

Instead, you keep improving gradually.

That is Online Learning.

---

## Working Process

```
New Data
    │
    ▼
Update Model
    │
    ▼
Better Model
    │
    ▼
New Data
    │
    ▼
Update Again
```

---

## Example

Consider a news recommendation system.

Every time a user:

- clicks an article,
- likes a story, or
- watches a video,

the recommendation model updates itself to better match that user's interests.

---

## Characteristics

- Learns continuously.
- Suitable for streaming data.
- Low memory usage.
- Can adapt to changing patterns.
- Supports real-time learning.

---

## Advantages

- Learns from new data immediately.
- No need to retrain from scratch.
- Low memory requirement.
- Handles continuously changing environments.

---

## Disadvantages

- Sensitive to noisy or incorrect data.
- Can gradually forget older patterns (called **catastrophic forgetting** in some cases).
- Harder to monitor and debug.
- Requires careful tuning of learning rate and update strategy.

---

## Applications

- Stock Price Prediction
- Fraud Detection
- Recommendation Systems
- Spam Detection
- Social Media Feed Ranking
- IoT Sensor Data
- Online Advertisement Systems

---

# 3. Out-of-Core Learning

## Definition

**Out-of-Core Learning** is a learning approach designed for **very large datasets that cannot fit into the computer's RAM**.

Instead of loading the entire dataset into memory, the data is divided into **small chunks (batches)**.

The model processes one chunk at a time until the entire dataset has been used.

> **Simple Definition:**  
> Out-of-Core Learning trains on very large datasets by loading only small portions into memory.

---

## Why is Out-of-Core Learning Needed?

Suppose your computer has:

```
RAM = 8 GB
```

But your dataset size is:

```
500 GB
```

It is impossible to load all 500 GB into memory at once.

Solution:

Load:

```
Chunk 1 → Train

Chunk 2 → Train

Chunk 3 → Train

...

Continue until all chunks are processed.
```

This is Out-of-Core Learning.

---

## Working Process

```
Huge Dataset
      │
      ▼
Divide into Small Chunks
      │
      ▼
Load Chunk 1
      │
      ▼
Train Model
      │
      ▼
Load Chunk 2
      │
      ▼
Update Model
      │
      ▼
Repeat Until Finished
```

---

## Example

Suppose we have:

```
1 TB Dataset
```

Computer RAM:

```
16 GB
```

The dataset is divided into:

```
Chunk 1 = 2 GB

Chunk 2 = 2 GB

Chunk 3 = 2 GB

...

Chunk 500
```

The model processes one chunk at a time without loading the full dataset into memory.

---

## Characteristics

- Handles datasets larger than RAM.
- Uses disk storage efficiently.
- Processes data incrementally.
- Often combined with Online Learning algorithms.

---

## Advantages

- Can train on extremely large datasets.
- Low memory usage.
- Scalable to big data.
- Suitable for distributed systems.

---

## Disadvantages

- Slower than in-memory training because reading from disk is slower than RAM.
- More complex implementation.
- Performance depends on disk speed.

---

## Applications

- Big Data Analytics
- Large Image Datasets
- Social Media Data Analysis
- Search Engines
- Scientific Data Processing
- Large Log File Analysis

---

# Batch Learning vs Online Learning vs Out-of-Core Learning

| Feature | Batch Learning | Online Learning | Out-of-Core Learning |
|---------|----------------|-----------------|----------------------|
| Training Data | Entire dataset | One sample or small batches | Small chunks of a huge dataset |
| Learning Style | Train once | Continuous updates | Incremental chunk-by-chunk |
| Memory Requirement | High | Low | Low |
| Handles New Data | ❌ Retraining required | ✅ Yes | ✅ Yes (by processing additional chunks) |
| Training Speed | Slow | Fast updates | Moderate (depends on disk speed) |
| Suitable for Big Data | Limited | Yes | Excellent |
| Real-Time Learning | ❌ No | ✅ Yes | Usually No (focus is large data, not live streams) |

# When Should We Use Each?

### Use Batch Learning when:

- Dataset fits into RAM.
- Data rarely changes.
- Maximum accuracy is desired.

---

### Use Online Learning when:

- Data arrives continuously.
- The environment changes over time.
- Real-time updates are required.

---

### Use Out-of-Core Learning when:

- Dataset is larger than available RAM.
- Working with Big Data.
- Memory is limited.

# Interview / Exam Points

- **Batch Learning** trains the model using the complete dataset at once and requires retraining when new data is added.
- **Online Learning** continuously updates the model with new data, making it suitable for streaming and real-time applications.
- **Out-of-Core Learning** processes very large datasets in small chunks because the full dataset cannot fit into RAM.
- Batch Learning requires high memory, while Online and Out-of-Core Learning are more memory-efficient.
- Out-of-Core Learning is primarily designed to overcome memory limitations, whereas Online Learning is designed for continuous learning from incoming data.

---

# Summary

Machine Learning models can learn in different ways depending on the nature of the data. **Batch Learning** trains once using the entire dataset and requires retraining when new data becomes available. **Online Learning** updates the model continuously as new data arrives, making it ideal for real-time systems. **Out-of-Core Learning** enables training on datasets larger than the available memory by processing the data in small chunks, making it an essential technique for big data applications.

````python
# This code is for Online Learning.
import numpy as np
from sklearn import linear_model

n_samples, n_features = 1, 500

y = np.random.randn(n_samples)
X = np.random.randn(n_samples, n_features)

clf = linear_model.SGDRegressor()

import time

start_time = time.time()

clf.partial_fit(X, y)

elapsed_time = time.time() - start_time

print(elapsed_time)
````

````python
print("Sum of weights before second fit:", np.sum(clf.coef_))
````

````python
y = np.random.randn(n_samples)
X = np.random.randn(n_samples, n_features)
````

````python
clf.partial_fit(X, y)
````

````python
print("Sum of weights after second fit: ", np.sum(clf.coef_))
````
