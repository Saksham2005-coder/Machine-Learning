# Tensors in Machine Learning

## What are Tensors?

A **Tensor** is a mathematical object used to represent data in Machine Learning and Deep Learning.

A tensor is simply a **container for numbers** arranged in one or more dimensions.

You can think of a tensor as a **generalization of scalars, vectors, and matrices**.

- A **Scalar** is a 0D Tensor.
- A **Vector** is a 1D Tensor.
- A **Matrix** is a 2D Tensor.
- Higher-dimensional arrays are called **3D, 4D, 5D, ... ND Tensors**.

> **Simple Definition:**  
> A Tensor is a multi-dimensional array used to store and process data in Machine Learning and Deep Learning.

---

# Why are Tensors Important?

Deep Learning libraries such as:

- TensorFlow
- PyTorch
- JAX
- MXNet

store all data in the form of tensors.

Examples of data represented as tensors:

- Images
- Videos
- Audio
- Text
- Sensor data
- Medical records

Every input, output, weight, and intermediate computation inside a neural network is represented as a tensor.

# Interview / Exam Points

- A **Tensor** is a multi-dimensional array used to represent data in Machine Learning and Deep Learning.
- **Scalar**, **Vector**, and **Matrix** are examples of **0D**, **1D**, and **2D** tensors, respectively.
- **Rank** is the number of dimensions (axes) of a tensor.
- **Shape** describes the size of the tensor along each axis.
- **Axis** represents one dimension of the tensor.
- Images are commonly represented as **3D tensors** (`Height × Width × Channels`).
- A batch of images is represented as a **4D tensor**, while a batch of videos is typically represented as a **5D tensor**.

---

# Summary

A **Tensor** is the fundamental data structure used in Machine Learning and Deep Learning. It generalizes scalars, vectors, and matrices to multiple dimensions, enabling efficient storage and computation of complex data. The **rank** indicates the number of dimensions, the **shape** specifies the size along each dimension, and the **axes** define the directions of those dimensions. Tensors are widely used to represent numerical data such as images, videos, text embeddings, and neural network parameters, making them the foundation of modern AI frameworks like TensorFlow and PyTorch.
