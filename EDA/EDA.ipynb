# Understanding Your Data (Exploratory Data Analysis - Initial Data Exploration)

## Introduction

Before building any Machine Learning model, it is essential to **understand the dataset**.

This is the **first stage of Exploratory Data Analysis (EDA)**, where we inspect the data to understand its structure, quality, and characteristics.

Understanding the data helps us:

- Detect errors
- Identify missing values
- Find duplicate records
- Understand relationships between variables
- Decide preprocessing steps
- Select appropriate Machine Learning algorithms

> **Simple Definition:**  
> Understanding Your Data means exploring the dataset to know what information it contains and whether it is suitable for Machine Learning.

---

# Why is Understanding Data Important?

Imagine a doctor treating a patient without checking:

- Symptoms
- Medical history
- Test reports

The treatment would likely fail.

Similarly, building an ML model without understanding the data usually leads to poor results.

---

# Questions We Should Ask About Our Data

Whenever a new dataset is received, ask the following questions:

1. How big is the data?
2. How does the data look?
3. What are the data types of the columns?
4. Are there any missing values?
5. How does the data look mathematically?
6. Are there any duplicate values?
7. How are the columns correlated?

These questions form the foundation of **Exploratory Data Analysis (EDA).**

# Univariate Analysis

## Introduction

**Univariate Analysis** is one of the most basic and important steps in **Exploratory Data Analysis (EDA)**.

The word **Univariate** can be broken into:

- **Uni** = One
- **Variate** = Variable (Feature/Column)

So, Univariate Analysis means analyzing **only one variable at a time**.

Its main purpose is to understand:

- Distribution of data
- Frequency of values
- Central tendency
- Spread of data
- Presence of outliers
- Shape of the data

> **Simple Definition:**  
> Univariate Analysis is the process of analyzing a single variable independently to understand its characteristics.

---

# Why is Univariate Analysis Important?

It helps us answer questions like:

- Which category occurs the most?
- Is the data normally distributed?
- Are there any outliers?
- What is the range of values?
- Is the data skewed?
- What is the average value?

These insights help in:

- Data cleaning
- Feature engineering
- Feature selection
- Choosing suitable ML algorithms

---

# Univariate Analysis Workflow

```
Choose One Column
        │
        ▼
Identify Data Type
        │
        ▼
Select Appropriate Plot
        │
        ▼
Analyze Distribution
        │
        ▼
Find Outliers
        │
        ▼
Draw Conclusions
```

# Boxplot

## Definition

A **Boxplot** summarizes the distribution of numerical data using the **Five-Number Summary**.

It is the best plot for detecting **outliers**.

> **Simple Definition:**  
> A Boxplot shows the spread of data and highlights unusual values (outliers).

---

## Five-Number Summary

A Boxplot is based on:

- Minimum
- First Quartile (Q1)
- Median (Q2)
- Third Quartile (Q3)
- Maximum

---

## Structure of a Boxplot

```
Minimum      Maximum
   │             │
   ▼             ▼

----|====|====|----

    Q1  Median Q3

      □ = Middle 50%
```

---

## Example

Marks

```
45

50

52

55

60

62

65

70

150
```

```
150
```

is an **outlier**.

The Boxplot will display it as a separate point.

---

## Seaborn Code

```python
sns.boxplot(x=df["Marks"])
```

---

## Advantages

- Detects outliers.
- Shows spread of data.
- Compares multiple distributions.

---

## Disadvantages

- Does not show every individual observation.
- Less intuitive for beginners than histograms.

---

## Applications

- Salary Analysis
- Marks Analysis
- Medical Data
- Fraud Detection
- Quality Control

# Interview / Exam Points

- **Univariate Analysis** studies **one variable at a time**.
- Data is broadly classified into **Categorical** (Nominal, Ordinal) and **Numerical** (Discrete, Continuous).
- **Countplot** displays the frequency of each category and is used for categorical data.
- **Pie Chart** shows the percentage contribution of each category and is suitable for a small number of categories.
- **Distplot** (now commonly replaced by `histplot()` with `kde=True`) is used to visualize the distribution of numerical data.
- **Boxplot** displays the five-number summary and is particularly useful for detecting outliers.

# Summary

Univariate Analysis is the first step in Exploratory Data Analysis (EDA), where a single variable is analyzed independently. The choice of visualization depends on the data type: **Countplots** and **Pie Charts** are ideal for categorical variables, while **Histograms (Distplots)** and **Boxplots** are used for numerical variables. These visualizations help understand distributions, identify patterns, detect outliers, and prepare the data for further analysis and Machine Learning.

# Exploratory Data Analysis (EDA) using Bivariate and Multivariate Analysis

## Introduction

After understanding individual variables using **Univariate Analysis**, the next step in Exploratory Data Analysis (EDA) is to study the **relationship between two or more variables**.

This helps answer questions like:

- Does salary increase with experience?
- Does study time affect marks?
- Does age influence loan approval?
- Which features are highly correlated?

These analyses help us discover relationships that improve Machine Learning models.

> **Simple Definition:**

Analyze

# Interview / Exam Points

- **Bivariate Analysis** examines the relationship between **two variables**, while **Multivariate Analysis** examines **three or more variables** simultaneously.
- The choice of visualization depends on the data types:
  - **Numerical vs Numerical:** Scatter Plot, Correlation
  - **Numerical vs Categorical:** Boxplot, Bar Plot
  - **Categorical vs Categorical:** Crosstab, Heatmap
- **Pair Plots** help visualize relationships between multiple numerical features.
- **Heatmaps** display correlation matrices and are useful for feature selection and detecting multicollinearity.
- Bivariate and Multivariate Analysis are important for identifying relationships, selecting relevant features, and improving Machine Learning model performance.

# Summary

Bivariate and Multivariate Analysis are essential stages of Exploratory Data Analysis (EDA). **Bivariate Analysis** explores the relationship between two variables using tools such as scatter plots, boxplots, bar plots, and crosstabs. **Multivariate Analysis** extends this concept to three or more variables, using techniques like pair plots, heatmaps, grouped boxplots, and 3D scatter plots to uncover complex interactions. These analyses provide valuable insights into feature relationships, support feature selection, and help build more accurate and reliable Machine Learning models.
